```python
s = input("Enter string: ")
letters = ""
total = 0
for c in s:
    if c.isalpha():
        letters += c
    elif c.isdigit():
        total += int(c)
print(f'Word: "{letters}", Sum: {total}')
```

```python
p = input("Enter password: ")
has_upper = False
has_special = False
for c in p:
    if c.isupper():
        has_upper = True
    if c in "@#$%&*!":
        has_special = True
if 8 <= len(p) <= 16 and has_upper and has_special:
    print("Password is Secure.")
else:
    print("Password is not secure.")
```

```python
price = float(input("Price: "))
member = input("Member (Yes/No): ").lower()
if price > 100:
    price = price * 0.8
elif price >= 50:
    price = price * 0.9
if member == "yes":
    price = price * 0.95
print("Final Total: $", round(price, 2))
```

```python
students = {'Sam': [95, 90, 92], 'Ali': [70, 80, 75], 'Sara': [98, 95, 99]}
for name in students:
    avg = sum(students[name]) / 3
    if avg > 90:
        print(f"{name}: Average {avg:.1f} - Honor Roll")
    else:
        print(f"{name}: Average {avg:.1f}")
```

```python
n = input("Number: ")
while len(n) > 1:
    total = 0
    for d in n:
        total += int(d)
    n = str(total)
print("Digital Root:", n)
```

```python
nums = [1, 1, 2, 3, 3, 3, 4]
seen = []
result = []
for x in nums:
    if x not in seen:
        seen.append(x)
        result.append((x, nums.count(x)))
print(result)
```

```python
n = int(input("n: "))
f = 1
s = 0
for i in range(1, n + 1):
    f *= i
    s += i
if f % s == 0:
    print(f"Factorial is {f}, Sum is {s}. {f} is divisible by {s}.")
else:
    print(f"Factorial is {f}, Sum is {s}. {f} is not divisible by {s}.")
```

```python
store = {'apple': 0.5, 'bread': 2.0, 'milk': 1.5}
cart = []
while True:
    action = input("Add / Remove / Done: ").lower()
    if action == "done":
        break
    item = input("Item: ").lower()
    if action == "add":
        if item in store:
            cart.append(item)
        else:
            print("Error: item not in store")
    elif action == "remove":
        if item in cart:
            cart.remove(item)
total = 0
for item in cart:
    total += store[item]
print("Total Bill: $", total)
```

```python
nums = [1, 5, 2, 6, 3]
peaks = []
for i in range(1, len(nums) - 1):
    if nums[i] > nums[i - 1] and nums[i] > nums[i + 1]:
        peaks.append(nums[i])
print("Peaks:", peaks)
```

```python
d = {'A': 1, 'B': 2, 'C': 1}
result = {}
for k in d:
    v = d[k]
    if v in result:
        result[v].append(k)
    else:
        result[v] = [k]
print(result)
```
